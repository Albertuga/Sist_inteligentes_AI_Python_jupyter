```python
from copy import deepcopy
from typing import Tuple, List
from IPython.display import display
from ipywidgets import HTML
import math
import random

class Tablerowumpus:
    
    def __init__(self, size):
        
        #solicitamos tamaño al usuario, directamente desde el argumento
        self.size = size
        #adaptamos tamaño de matrix
        self.matrix = self.inicializar_tablero()
        self.previous_content = "B" #guardar el contenido original de cada celda antes de mover
        self.agent_pos = self.pos_agente()
        self.game_over = False #bandera que nos indica si ha acabdo el juego
        self.flecha_disponible=True #valoramos si hay flechas disponibles
        self.posiciones_visitadas =set()
        self.posiciones_hedor = set()
        
     # Función para asociar cada carácter a un número del tablero
    def getCodigoTablero(self, caracter):
        codigo = 0

        if caracter == 'A':  # Agente
            codigo = 1
        elif caracter == 'W':  # Wumpus
            codigo = 2        
        elif caracter == 'H':  # Hedor
            codigo = 3
        elif caracter == 'O':  # Oro
            codigo = 4
        elif caracter == 'S':  # Brisa
            codigo = 5
        elif caracter == 'C':  # Hueco
            codigo = 6            
        return codigo

    # Asocia el código obtenido con un nombre para la imagen
    def get_content(self, coord, tablero):
        codigo = self.getCodigoTablero(tablero[coord[0]][coord[1]])
        contenidos = {
            0: "CasillaVacia",
            1: "CasillaAgente",
            2: "CasillaWumpus",
            3: "CasillaHedor",
            4: "CasillaOro",
            5: "CasillaBrisa",
            6: "CasillaHueco"
        }
        return contenidos[codigo]

    # Precargamos las rutas de las imágenes del tablero
    element_image = {
        "CasillaVacia": "./ImagenesCasillasWumpus/CasillaVacia.png",
        "CasillaAgente": "./ImagenesCasillasWumpus/CasillaAgente.png",
        "CasillaWumpus": "./ImagenesCasillasWumpus/CasillaWumpus.png",
        "CasillaHedor": "./ImagenesCasillasWumpus/CasillaHedor.png",
        "CasillaOro": "./ImagenesCasillasWumpus/CasillaOro.png",
        "CasillaBrisa": "./ImagenesCasillasWumpus/CasillaBrisa.png",
        "CasillaHueco": "./ImagenesCasillasWumpus/CasillaHueco.png"
    }
    
    # Función que muestra una representación gráfica del tablero en HTML
    def get_html(self, tablero):
        """Muestra una representación gráfica del juego en HTML."""
        height = len(tablero)
        width = len(tablero[0])

        html_string = "<style> img.game {width: 37px !important; height: 22px !important;}</style><table>"
        for i in range(height):
            html_string += "<tr>"
            for j in range(width):
                content = self.get_content((i, j), tablero)  #Obtener el contenido de la celda
                image_path = self.element_image[content]     #Obtener la ruta de la imagen
                html_string += f'<td><img class="game" src="{image_path}" alt=""></td>'
            html_string += "</tr>"
        html_string += "</table>"

        return html_string

    # Inicializa el tablero con contenido aleatorio
    def inicializar_tablero(self):
        global TableroIni
        #inicializamos un tablero vacio donde colocar fichas
        tablero = [["B" for _ in range(self.size)] for _ in range(self.size)]  # 'B' para celdas vacías
        
        #solicitamos cantidad de huecos al usuario
        while True:
            num_huecos=int(input("Escoja la cantidad de huecos para la partida (máximo 3 huecos)"))
            #vamos a pedir que solo se puedan agregar hasta 3 huecos (pero el juego funciona incluso con mas si quitamos esto)
            if 0<=num_huecos<=3: #podemos jugar sin huecos si quiere
                break
            else:
                print("Debe introducir un numero entre 0 a 3.")
                
        fichas = ['W', 'O'] + ['C'] * num_huecos  #Agente, Wumpus, Oro y dos huecos
        agente_pos = (self.size -1, 0)  #Inicializamos el agente en su posicion inicial
        wumpus_pos = None   #Inicializamos el wumpus para hacer restricciones necesarias
        
        #colocamos al agente en la esquina inf izquierda
        tablero[agente_pos[0]][agente_pos[1]] = "A"
        
        #Coloca las fichas sin repetición de posiciones
        posiciones_usadas = set()
        posiciones_usadas.add(agente_pos) #evitamos que wumpus, hueco y oro usen la posicion del agente
       
        for ficha in fichas:
            while True:
                fila = random.randint(0, self.size - 1)  #Distancias aleatorias en filas
                columna = random.randint(0, self.size - 1)  #Distancias aleatorias en columnas
                if (fila, columna) not in posiciones_usadas and (fila, columna) not in agente_pos: 
                    #Evitamos que agente y oro estén juntos
                    if ficha == "O" and self.esAdyacente(agente_pos, (fila, columna)):
                        continue  #Intentamos otra posición en caso que sean adyacentes
                    elif ficha == "W" and self.esAdyacente(agente_pos, (fila, columna)):
                        continue
                    elif ficha == "C" and self.esAdyacente(agente_pos, (fila, columna)):
                        continue
                    elif ficha == "A":
                        agente_pos = (fila, columna)
                #Añadimos otra excepcion para que el wumpus no este continuo a huecos
                    if ficha == "C" and wumpus_pos and self.esAdyacente(wumpus_pos, (fila, columna)):
                        continue
                    elif ficha == "W":
                        wumpus_pos = (fila, columna)
                    
                    posiciones_usadas.add((fila, columna))
                    tablero[fila][columna] = ficha
                    break

        #Añadimos hedor y brisa alrededor del Wumpus y los huecos
        for i in range(self.size):
            for j in range(self.size):
                if tablero[i][j] == 'W':  #Si hay un Wumpus en (i, j)
                    for x, y in [(i - 1, j), (i + 1, j), (i, j - 1), (i, j + 1)]:
                        #añadimos alrededor del wumpus el hedor en las cledas vacias
                        if 0 <= x < self.size and 0 <= y < self.size and tablero[x][y] == 'B':
                            tablero[x][y] = 'H'
                elif tablero[i][j] == 'C':  # Si hay un hueco en (i, j)
                    for x, y in [(i - 1, j), (i + 1, j), (i, j - 1), (i, j + 1)]:
                        if 0 <= x < self.size and 0 <= y < self.size and tablero[x][y] == 'B':
                            tablero[x][y] = 'S'
        
        TableroIni = tablero
        return tablero

    #funcion que nos indica si una posicion es adyacente a otra
    def esAdyacente(self, pos1, pos2):
        return abs(pos1[0] - pos2[0]) + abs(pos1[1] - pos2[1]) == 1

    
    def setMatrix(self, matrix):
        self.matrix = deepcopy(matrix)
    
    def getMatrix(self) -> List[List]:
        return deepcopy(self.matrix)
    
    def placeTile(self, row: int, col: int, tile: int):
        self.matrix[row][col] = tile
    
    #la funcion utility se encargará de calcular nuestro score dependiente de las distancias que tenemos desde el agente
    #hacia el oro, o wumpus, a la vez penaliza si estamos en brisas, hedores o voviendo a caer en posiciones ya visitadas
    #para evitar redundancia de movimientos
    def utility(self) -> float:
        agente_pos = self.pos_agente()
        oro_pos = self.gold_position()
        wumpus_pos = self.wumpus_position()

        if agente_pos == oro_pos:
            return math.inf  #Maximo coste si estan juntos
        if oro_pos is None or agente_pos is None:
            return math.inf #Maximo coste si desaparecen esto es lo que ocurre cuando gana

        score = 0.0

        #Recogemos la distancia euclidea del oro
        if oro_pos:
            distancia_oro = math.sqrt((agente_pos[0] - oro_pos[0])**2 + (agente_pos[1] - oro_pos[1])**2)
            score += max(0, 20000 - distancia_oro * 1000)#a medida que vamos acercandonos, vamos premiando al agente

        #recogemos la distancia euclidea del wumpus 
        if wumpus_pos:
            distancia_wumpus = math.sqrt((agente_pos[0] - wumpus_pos[0])**2 + (agente_pos[1] - wumpus_pos[1])**2)
            score -= max(0, 5000 - distancia_wumpus * 500) #a medida que nos vamos acercando, vamos penalizando

        #penalización por acercarse a huecos, brisas o hedor
        vecinos = self.identificaVecinos(agente_pos) #si son nuestras celdas adyacentes
        for vecino in vecinos:
            contenido = self.contiene(vecino)
            if contenido == 'C':  #Huecos
                score -= 5000
            elif contenido == 'S':  # Brisas
                score -= 1000
            elif contenido == 'H':  # Hedor
                score -= 2000

        # Penalización por revisitar celdas
        if agente_pos in self.posiciones_visitadas:
            score -= 20
        else:
            self.posiciones_visitadas.add(agente_pos)

        return score
 
    #funcion que retorna el contenido de la celda que entramos por argumento
    def contiene(self, coord):
        row, col = coord
        return self.matrix[row][col]

    def identificaVecinos(self, posicion):
        if posicion is None:
            return [] #retornamos lista vacia en caso de no encontrar pos
        fila, columna = posicion #recogemos [row] and  [col]
        vecinos = []

        #describimos las posibles direcciones
        direccion = [(-1, 0), (1, 0), (0, -1), (0, 1)]

        #calculamos cada vecino
        for dirf, dirc in direccion:
            nueva_fila = fila + dirf
            nueva_columna = columna + dirc

            #Verificar si la nueva posición está dentro de los límites del tablero
            if 0 <= nueva_fila < len(self.matrix) and 0 <= nueva_columna < len(self.matrix[0]):
                vecinos.append((nueva_fila, nueva_columna))

        return vecinos
    
    #para calcular los posibles movimientos de los huecos
    def possibleMoves(self, position: Tuple[int, int]) -> List[Tuple[int, int]]:
        #devolvemos lista de posiciones a la que se puede mover desde la pos actual
        movs_posibles = [(-1,0),(1,0),(0,-1),(0,1)] #Arriba, abajo izquieda derechao
        #array que contiene los movimientos validos
        movs_validos = []
        
        row, col = position
        for dx, dy in movs_posibles:
            n_row, n_col = row + dx, col + dy
            if self.cel_segura(n_row, n_col): #si la celda es segura
                movs_validos.append((n_row,n_col)) #agregamos el mov valido
                
        return movs_validos    
    
    #estos movimientos se hacen solo si estan dentro del tablero
    def canMoveUp(self,row: int, col: int) -> bool:
        return row > 0 #si estamos en la primera fila, no podremos movernos mas arriba
  
    def canMoveDown(self,row: int, col: int) -> bool:
         return row < self.size -1 #Verificamos que no este en la ultima fila
        
        
    def canMoveLeft(self,row: int, col: int) -> bool:
        return col > 0    #verificamos que si estamos en la primera col, no podremos salir del tablero


    def canMoveRight(self,row: int, col: int) -> bool:
         return col < self.size -1 #Verificamos si estamos en la ultima col para no salir del tablero
 

    #busca movimientos posibles que el agente  puede hacer desde una posicion
    def getAvailableMovesForMax(self,row: int, col: int) -> List[int]:
        max_moves = [] #lista que contiene los movimientos max
        
        direccion= [(-1,0),(1,0),(0,-1),(0,1)]
        
        for dirx, diry in direccion:
            Nrow, Ncol = row + dirx, col +diry
            if self.cel_segura(Nrow, Ncol):#verificamos que la celda sea segura
                max_moves.append((Nrow, Ncol))
        
        return max_moves #retornamos lista
    
    #busca los movminientos del jugador min que puede representar el hueco aleatorio
    def getAvailableMovesForMin(self) -> List[Tuple[int]]:
        min_moves = []
        
        #iteramos sobre el tablero buscando elementos min
        for i in range(len(self.matrix)):
            for j in range(len(self.matrix[0])):
                if self.matrix[i][j] == 'C': #solo movemos los huecos!
                    posicion = (i,j)
                    #obtenemos movimientos para cada posicion de MIN
                    moves = self.getAvailableMovesForMax(i,j) #reutilizamos logica de movimiento
                    for nuevo_pos in moves:
                        min_moves.append((posicion, nuevo_pos))
                
        return min_moves
    
    #funcion que permite detectar al wumpus si esta adyacente al agente y luego dispara
    def detectarWumpus(self):
        #si el agente no esta 
        if self.pos_agente() is None:
            return #retorna nada (eliminamos errores)
        fila, columna = self.pos_agente()#posiciones del agente
        posiciones_hedor = [] #array de hedores..

        #identificar las posiciones con hedor alrededor del agente
        for dirx, diry in [(-1, 0), (1, 0), (0, -1), (0, 1)]: 
            Nrow, Ncol = fila + dirx, columna + diry #agregamos esta ubicacion a nuestra pos agente
            if 0 <= Nrow < self.size and 0 <= Ncol < self.size: #verificamos que esten dentro de tablero
                if self.matrix[Nrow][Ncol] == 'H': #y que contengan hedor
                    posiciones_hedor.append((Nrow, Ncol)) #anexamos a nuestro array de hedores...

        if not posiciones_hedor: #si no hay ningun hedor no agregamos
            return

        print(f"Posiciones con hedor detectadas: {posiciones_hedor}")

        #si detecta el hedor, evalua si disparar o moverse
        for dirx, diry in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
            Nrow, Ncol = fila + dirx, columna + diry #agregamos esta ubicacion a nuestra pos agente
            if (Nrow, Ncol) in posiciones_hedor: #verificamos si esta 
                #confirmar si el Wumpus está en la dirección del hedor
                if self.percepcion_wumpus(fila, columna, dirx, diry):
                    print(f"Disparando flecha en dirección {self.get_direccion(dirx, diry)}")
                    resultado = self.dispara_flecha(self.get_direccion(dirx, diry))
                    print(resultado)
                    return
        print("El agente decide moverse hacia las posiciones con hedor.")
    
    #auxiliar: se utiliza para disparar hacia la direccion que el agente intuye que esta el wumpus
    def percepcion_wumpus(self, x, y, dx, dy):
        #limitamos tablero
        #x,y =row and col del agente
        while 0 <= x < self.size and 0 <= y < self.size:  #Mientras esté dentro del tablero
            if self.matrix[x][y] == 'W':
                return True  #Confirmamos Presencia del Wumpus!

            #incrementamos coordenadas
            #prevenir error out of range!
            x += dx
            y += dy

            # Verificar si estamos dentro de los límites después de incrementar
            if not (0 <= x < self.size and 0 <= y < self.size):
                break #salir si esta fuera del tablero

        return False #no hay wumpus

    
    #funcion para dar direccion al disparo (recorre camino de fila o columna)
    #usado solo para disparar en modo automatico, manteniendo el mismo metodo dispara_flecha
    def get_direccion(self, dirx, diry):
        #desplazamiento se convierte en direccion
        if dirx==-1: return "w"#Arriba
        if dirx==1: return "s" #Abajo
        if diry==-1: return "a"#Izquierda
        if diry==1: return "d" #Derecha
    
    def dispara_flecha(self, direccion: str):
        #vemos si ya la hemos utilizado
        if not self.flecha_disponible:
            return "Ya usaste la flecha, no tienes más flechas."
        #obtenemos coordenadas del agente
        agente_row, agente_col = self.pos_agente()
        
        #establecemos las direcciones en que se puede disparar
        mov = {
            "w":(-1,0), #arriba
            "s":(1,0), #abajo
            "d":(0,1), #derecha
            "a":(0,-1) #izquierda
        }
        
        #aplicamos excepcion si el movimiento no es valido
        if direccion not in mov:
            return "dirección invalida para el disparo."
        
        #indicamos recorrido de la flecha
        d_row, d_col = mov[direccion]
        
        #recorrido de flechas
        row, col, = agente_row, agente_col
        while 0 <= row < self.size and 0 <= col < self.size:
            #vemos si la flecha impacta en el wumpus
            if self.matrix[row][col] == "W":
                self.matrix[row][col] = "B" #elimina wumpus y pone celda en blanco
                self.flecha_disponible = False # hemos usado la flecha
                #Eliminamos el hedor del wumpus
                for x, y in [(row - 1, col), (row + 1, col), (row, col - 1), (row, col + 1)]:
                    if 0 <= x < self.size and 0 <= y < self.size and self.matrix[x][y] == 'H':
                        self.matrix[x][y] = 'B'
                return "¡Has eliminado al Wumpus!"
            
            #Avanza en la direccion que hemos disparado
            row+=d_row
            col+=d_col
        self.flecha_disponible = False    
        return "La flecha no impacto en el Wumpus."
    
    #funcion que retorna la posicion del agente
    def pos_agente(self):
        #encontramos la posicion del agente 'A' en el tablero
        for i in range(self.size):
            for j in range(self.size):
                if self.matrix[i][j]=='A':
                    return i,j
        return None
    
    #funcion que retorna posicion del oro
    def gold_position(self):
        for i in range(len(self.matrix)):
            for j in range(len(self.matrix[0])):
                if self.matrix[i][j] == 'O':
                    return i, j
        return None
    
    #funcion que retorna la posicion del wumpus
    def wumpus_position(self):
        for i in range(len(self.matrix)):
            for j in range(len(self.matrix[0])):
                if self.matrix[i][j] == 'W':
                    return i, j
        return None
    
    #funcion que verifica si la celda es segura o no
    def cel_segura(self, row, col) -> bool:
        #ahora verificamos si la celda a la que nos vamos a mover esta en los limites
            if 0 <= row < self.size and 0 <= col < self.size:
                contenido = self.matrix[row][col]
                #verificamos contenido de celda pasada por parametro
                #esta celda es segura si no contiene wumpus ni hueco
                if contenido not in ['W', 'C']:
                    return True
            return False
                
    def moveCanBeMade(self,player: int) -> bool:
        if self.pos_agente() is None:
            return False
        
        #diseñamos una lista con posibles movimientos del tablero
        row, col = self.pos_agente()
        posibles_moves = [(-1,0),(1,0),(0,-1),(0,1)] #Arriba,abajo,izquierda,derecha
        
        
        #itermaos en estos movimientos
        for dirx, diry in posibles_moves:
            new_fila, new_col = row + dirx, col + diry
            #si la celda no contiene wumpus ni hueco entonces true si no false
            if self.cel_segura(new_fila, new_col):
                return True #hay un mov valido
        return False #no hay mov valido
    

    
    #funcion que implementa todo el movimiento aleatorio de los huecos
    #selecciona un hueco aleatoriamente y lo mueve aleatoriamente despues de cada jugada del agente
    def moverHuecoAleatorio(self):
        #iteramos en busqueda de todos los huecos del tablero
        huecos = [(i, j) for i in range(self.size) for j in range(self.size) if self.matrix[i][j] == "C"]

        if not huecos:
            return False  #si no hay entonces retornamos 

        #Seleccionamos hueco aleatoriamente
        hueco = random.choice(huecos)
        row, col = hueco  #obtenemos ubicacion

        #definimos posibles movimientos
        movimientos_posibles = self.possibleMoves((row, col))

        #si no hay movimientos posibles
        if not movimientos_posibles:
            return False #no hacemos nada

        #evitamos que los huecos aparezcan en celdas donde hay oro, hedor, wumpus o agente
        movimientos_validos = []
        for new_pos in movimientos_posibles: #iteramos las posibles posiciones nuevas
            to_row, to_col = new_pos #obtenemos direccion
            if self.matrix[to_row][to_col] not in ['O', 'H', 'W', 'A']:  #No podemos movernos a una celda con oro, wumpus, agente o hedor
                movimientos_validos.append(new_pos) #agregamos nuestra lista de movimientos validos este filtrado de pos

        #si no hay mov validos
        if not movimientos_validos:
            return False   #no hacemos nada

        #Seleccionamos uno de estos movimientos
        new_pos = random.choice(movimientos_validos)
        to_row, to_col = new_pos #obtenemos direccion de estos movimientos

        #movemos el hueco a la nueva ubicacion
        self.matrix[row][col] = "B"  #casilla del hueco en blanco
        self.matrix[to_row][to_col] = "C"  #agregamos 'C' en la nueva ubicacion

        self.actualizarBrisasYHedores() #actualizamos brisas cada movimiento de huecos
        return True
    
    #funcion auxiliar que coloca y quita las brisas y hedores (que novuelvan a aparecer al pasar el agente)
    def actualizarBrisasYHedores(self):
        #eliminamos las brisas y hedores
        for i in range(self.size):#iteramos por el tablero
            for j in range(self.size):
                if self.matrix[i][j] in ['S', 'H']: #vaciamos celdas con brisa y hedor
                    self.matrix[i][j] = 'B'  #devolvemos celdas vacias

        #Añadimos las brisas alrededor de los huecos
        for i in range(self.size): #iteramos el tablero
            for j in range(self.size):
                if self.matrix[i][j] == 'C':  #Encontramos todos los huecos
                    for dx, dy in [(-1, 0), (1, 0), (0, -1), (0, 1)]:  #cogemos direcciones adyacentes
                        nx, ny = i + dx, j + dy #obtenemos estas ubicaciones y vemos que esten dentro del tablero
                        if 0 <= nx < self.size and 0 <= ny < self.size and self.matrix[nx][ny] == 'B':
                            self.matrix[nx][ny] = 'S'  #añadimos brisa

        #añadimos hedores con el mismo metodo (evitando que se borren al pasar huecos o agente encima)
        for i in range(self.size):#iteramos el tablero
            for j in range(self.size):
                if self.matrix[i][j] == 'W':  #encontramos al Wumpus
                    for dx, dy in [(-1, 0), (1, 0), (0, -1), (0, 1)]:  #direcciones adyacentes
                        nx, ny = i + dx, j + dy
                        if 0 <= nx < self.size and 0 <= ny < self.size and self.matrix[nx][ny] == 'B':
                            self.matrix[nx][ny] = 'H'  #añadimos hedor


    def isGameOver(self, row: int , col: int) -> bool:
        #en caso de que el agente consiga el oro
        if self.matrix[row][col] == "O":
            print("Has encontrado el oro, ¡GANASTE!")
            return True
        
        #verificamos si el agente se topa con el wumpus o hueco
        elif self.matrix[row][col] == 'W':
            print("¡Te ha capturado el Wumpus! Game Over")
            return True
        elif self.matrix[row][col] == 'C':
            print("¡Has caido en un Hueco! Game Over")
            return True
        elif self.pos_agente() is None:
            print("Has perdido, el agente ha caido en un hueco Game Over")
            return True
        #si aun no ha acabado la partida entonces
        return False

    def mover_agente(self, direccion):
    #obtenemos row and col del agent_pos
        row, col = self.agent_pos
        row1, col1 = row, col #duplicamos estas variables (ver automatica)
        nueva_pos = None
    #definimos direcciones
        if direccion == 'w':  #Arriba
            nueva_pos = (row - 1, col)
        elif direccion == 's':  #Abajo
            nueva_pos = (row + 1, col)
        elif direccion == 'a':  #Izquierda
            nueva_pos = (row, col - 1)
        elif direccion == 'd':  #Derecha
            nueva_pos = (row, col + 1)
        else: 
            nueva_pos = direccion
            row1=row
            col1 =col
            
        #Verificar si la nueva posición está dentro de los límites
        if nueva_pos and 0 <= nueva_pos[0] < self.size and 0 <= nueva_pos[1] < self.size:
            
            #GUARDAMOS CONTENIDO DE NUEVA POSICION
            row, col = nueva_pos
            self.previous_content = self.matrix[row][col]
 
            #Verificar si el juego ha terminado
            self.game_over = self.isGameOver(row, col)
            if not self.game_over:
                #movemos agente a nueva posicion
                self.matrix[row][col] = 'A'  # Mueve el agente a la nueva posición
                self.agent_pos = nueva_pos
                ultimoMov = direccion
            #volvemos a dar el valor a la casilla por donde el agente pasa
            #tomamos en cuenta el tablero inicial y actualizamos el hedor o brisa que tenia
            if TableroIni[row1][col1] in ['H', 'S']:
                self.matrix[row1][col1] = TableroIni[row1][col1]
            else:
                self.placeTile(row1, col1, 'B')
    
#Función que solicita al usuario el tamaño del tablero
def elegir_tamano_tablero():
    while True:
        try:
            tamano = int(input("Elige el tamaño del tablero (4 para 4x4, 6 para 6x6, 8 para 8x8): "))
            if tamano in [4, 6, 8]:  
                return tamano
            print("Opción inválida. Por favor, elige 4 o 6.")
        except ValueError:
            print("Entrada inválida. Por favor, ingresa un número.")
        
```


## Función asociada al método *Minimax* (con poda $\alpha-\beta$)

```python
def miniMax(state:Tablerowumpus, currentLevel:int, maxLevel:int, player:int, ultimoMov, alpha:int, beta:int, stop:bool) -> Tuple[Tablerowumpus, int, bool]:

    matriz=state.getMatrix()
        
    #si no hay mas movimientos posibles que el jugador puede hacer o llegamos a un maximo movimiento, entonces se detiene la exploracion
    if (not state.moveCanBeMade(player) or currentLevel == maxLevel):
        return (state.matrix,state.utility(),stop)  
    
    # ====================================================================================================
    #
    # A esta altura, si queréis podéis definir la fórmula de la función de coste, o llamarla.
    #
    # ====================================================================================================
          
    # =========================================================================================
    # Voy a generar el conjunto de los potenciales HIJOS/SUCESORES (en "successorMatrices"), y los voy a ir añadiendo
    # a "successormatrices".
    #
    # COMPLETAD, REHACED...COMO VEÁIS...
    #
    # La variable "successorBoard" bajo DEFINE UNO de los potenciales HIJOS.
    #
    # =========================================================================================           
    
    #definimos las matrices sucesoras
    successorMatrices = []
    
    #generamos sucesores que seran los movimientos validos del agente
    if player == 1: #MAX (agente)
        moves = state.getAvailableMovesForMax(*state.pos_agente()) #puntero para agregar [row] and [col]
        for move in moves:
            if ultimoMov != move:
                clon = deepcopy(state)
                posicion = state.pos_agente()
                clon.mover_agente(move) #movimiento del agente

                successorMatrices.append(clon) #obtenemos matrix del estado
    
    else: #'Otro jugador hueco'(va a realizar movimiento aleatorio de un hueco del tablero)
        matriz = state.getMatrix()
        #localizamos los huecos iterando por el tablero, obtenemos row and col
        huecos = [(i, j) for i in range(len(matriz)) for j in range(len(matriz[0])) if matriz[i][j] == "C"]
        for hueco in huecos:
            mov_valido = state.possibleMoves(hueco)
            for move in mov_valido:
                clon = deepcopy(state)
                clon.moverHuecoAleatorio() #IMPLEMENTA NUEVO MOV HUECO ALEATORIO
                successorMatrices.append(clon)
            
    #en caso de que no hayan sucesores (caminos) detener la exploracion
    if len(successorMatrices) == 0:
        stopDigging = True
        coste=state.utility()
        return (state.matrix,coste,stopDigging)
    
    #inicializar variables para encontrar el mejor resultado
    bestMatrix = None
    
    # En la siguiente línea, tenéis que definir adecuadamente el "IF".
    
    if player == 1: #turno del agente               
        maxValue = -math.inf #alpha
        
        for mat in successorMatrices:            
            #prefiero conservar el nombre de utility en lugar de best
                matrizS, utility, stop = miniMax(mat, currentLevel + 1, maxLevel, -1, ultimoMov, alpha,beta,stop)
                
                #print(f"Utilidad calculada para este estado: {utility}")
                maxValue = max(maxValue, utility)#actualizacion de alfa   
                if maxValue > alpha:
                    alpha = maxValue
                    bestMatrix = mat.getMatrix()
                    coste = maxValue
                    
                if alpha >= beta: #poda beta
                    break
            
    else:   #min (OTRO JUGADOR): Hueco              
        
        minValue = math.inf #beta
        for mat in successorMatrices:
            
                matrizS, utility, stop = miniMax(mat, currentLevel + 1, maxLevel, 1, ultimoMov, alpha,beta,stop)
            
                if utility < minValue:
                    minValue = utility
                    bestMatrix = mat.getMatrix()
                
            
                beta = min(beta, minValue) #actualizamos valor de beta
                if beta <= alpha: #poda alfa 
                    break

    return (bestMatrix,utility,stop)
            
```

```python
def performActionMinMax(state:Tablerowumpus, player:int, ultimoMov):
    
    matrizB=state.getMatrix()
        
    tmpMatriz = [row[:] for row in matrizB] 
        
    # ==================================================
    # Valor de profundidad que queráis. Yo recomiendo 2
    
    # depth = ........
    
    depth = 2
    matrizoptima = tmpMatriz
    stop = False 
    currentLevel = 0    
    
    state.setMatrix(tmpMatriz)
    
    #while not stop and itera <= 3:
    (best_move, minValue, stop) = miniMax(state, 0, 2, player, ultimoMov, -math.inf, math.inf, stop)       

    return best_move  
```

```python
def main():
    # Inicializar el tablero
    size = elegir_tamano_tablero() # Puedes cambiar esto o pedir al usuario que lo ingrese
    tablero = Tablerowumpus(size)
    ultimoMov = tablero.pos_agente()
    print("Tablero inicial:")
    display(HTML(tablero.get_html(tablero.matrix)))
    
    ultimo_estado= None
    
    # Ejecutar el algoritmo minimax
    while not tablero.game_over:
        
        # Mover el agente automáticamente usando minimax
        mejor_estado = performActionMinMax(tablero,1,ultimoMov) 
        ultimoMov = tablero.pos_agente()
            
        #Actualizamos el tablero y movemos al agente
        tablero.setMatrix(mejor_estado)
        tablero.mover_agente(tablero.pos_agente())
        tablero.actualizarBrisasYHedores()
        
        if tablero.pos_agente() is None:
            break
            
        #detectar y disparar al wumpus si es posible
        tablero.detectarWumpus() #llama dentro al disparo
        display(HTML(tablero.get_html(tablero.matrix))) #mostrar HTML    
        
        #Mover un hueco aleatoriamente después del movimiento del agente
        tablero.moverHuecoAleatorio()
        print("un Hueco se ha movido aleatoriamente")

if __name__ == "__main__":
    main()
```
